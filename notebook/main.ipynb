```python
# Penyesuaian Tipe Data
df['Order Date'] = pd.to_datetime(df['Order Date'])
df['Ship Date'] = pd.to_datetime(df['Ship Date'])
df['Sales'] = pd.to_numeric(df['Sales'])
df['Quantity'] = pd.to_numeric(df['Quantity'])
df['Discount'] = pd.to_numeric(df['Discount'])
df['Profit'] = pd.to_numeric(df['Profit'])
df.dtypes
```

```python
# Format Teks & Kode Pos
df["Postal Code"] = df["Postal Code"].astype(str).str.split(".").str[0].str.zfill(5)
```

```python
# Menghapus spasi berlebih
for c in df.select_dtypes(include=["object", "string"]).columns:
    df[c] = df[c].str.strip()
```

```python
# Mencari Outlier
def find_outliers_iqr(dataframe, column):
    Q1 = dataframe[column].quantile(0.25) 
    Q3 = dataframe[column].quantile(0.75)
    IQR = Q3 - Q1
    
    batas_bawah = Q1 - 1.5 * IQR
    batas_atas = Q3 + 1.5 * IQR
    
    outliers = dataframe[(dataframe[column] < batas_bawah) | (dataframe[column] > batas_atas)]
    return outliers, batas_bawah, batas_atas
```

```python
# outlier pada kolom Profit
outliers_profit, lower_profit, upper_profit = find_outliers_iqr(df, 'Profit')

# outlier pada kolom Sales
outliers_sales, lower_sales, upper_sales = find_outliers_iqr(df, 'Sales')

print(f"Total baris data keseluruhan: {len(df)}")
print(f"Jumlah outlier pada Profit: {len(outliers_profit)} baris")
print(f"Batas normal Profit: {lower_profit:.2f} s/d {upper_profit:.2f}\n")

print(f"Jumlah outlier pada Sales: {len(outliers_sales)} baris")
print(f"Batas normal Sales: {lower_sales:.2f} s/d {upper_sales:.2f}")
```

Pertahankan outlier dikarenakan semua baris tersebut adalah transaksi yang valid
